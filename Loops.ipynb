Program - The current population of a town is 10000.The population of town is increasing at the rate of 10% per year . You have to write a program to find out the population at the end of each of the last 10 years.

````python
## CODE HERE
curr_pop = 10000

for i in range(10,0,-1):
  print(i,curr_pop)
  curr_pop = curr_pop/1.1
````

**Sequence Sum**

1.   1/1! + 2/2! + 3/3! + .....



````python
## code here
n = int(input("Enter the number = "))
result = 0
fact = 1
for i in range(1,n+1):
  fact = fact * i
  result = result + (i/fact)

print(result)
````

**Pattern 1**



```
*
**
***
```














````python
## code here
n = int(input("Enter the number = "))
for i in range(1,n+1):
  for j in range(1,i+1):
    print('*',end='')
  print()
````

**PATTERN 2**


```
1
121
12321
1234321
```



````python
# CODE HERE
rows = int(input("Enter the rows = "))

for i in range(1,rows+1):
  for j in range(1,i+1):
    print(j,end="")
  for k in range(i-1,0,-1):
    print(k,end='') ## end = "\n"
  print()
````

**LOOPS Control Statement**


*   Break
*   Continue
*   Pass



````python
for i in range(1,6):
  if i == 3: ## stop at 2 not at 3 (Do not print 3)
    break ## Full break
  print(i)
````

````python
lower = int(input("Enter the lower number = "))
upper = int(input("Enter the Upper number = "))

for i in range(lower,upper+1):
  for j in range(2,i):
    if i%j == 0:
      break
  else:
    print(i)
````

````python
## Continue Statement
for i in range(1,11):
  if i == 5:
    continue ## Skip only 5
  print(i)
````

````python
## PASS
for i in range(1,10):
  pass ## Use for skip this code
````
