```python
## While loop example -> Program to print the table
number = int(input("Enter the number"))

i = 1 ## i = iteration
while i<11:
  print(number , "*" , i , "=",number*i)
  i+=1
```

```python
## While lopp with else
x =1
while x<3:
  print(x)
  x+=1

else:
  print("Limit cross")
```

Program - The current population of a town is 10000.The population of town is increasing at the rate of 10% per year . You have to write a program to find out the population at the end of each of the last 10 years.

```python
curr_pop = 10000

for i in range(10,0,-1):
  print(i,curr_pop)
  curr_pop = curr_pop - 0.1*curr_pop
```
