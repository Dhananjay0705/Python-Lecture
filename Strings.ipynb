**Capitalize/Title/Upper/Lower/Swapcase**

```python
s = 'hello world'
s.capitalize() ## sentence ke first character ko capitalize kar dta hai
```

```python
s.title() ## sentence ke every word ke first letter ko capitalize kar dta hai
```

```python
s.upper() ## sentence ke every word ko upper case me convert kar deta hai
```

```python
s.lower()  ## sentence ke every word ko lower case me convert kar deta hai
```

```python
'heLLo WoRlD'.swapcase() ## its convert lower to capital and capital convert into lower
```
